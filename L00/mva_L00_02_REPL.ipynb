## 実行順序の話

すべてのセルを実行した時には，上のセルから順に実行されます

この時点で代入されていない変数は値を持ちません

```python
b # エラーになる
```

```python
a
```

しかし，代入を一度実行した後で，上のセルを再クリックすると，今度は値を持ちます．つまり，セル内容の適用順序は，上からの順にではなく，実行された順（たまたま上からかもしれません）

```python
a=414
```

```python
a=123
```

`a=123`の直下にある下のセルですが，セルの実行の順序によっては，他の値に評価されたりします．

```python
a
```
