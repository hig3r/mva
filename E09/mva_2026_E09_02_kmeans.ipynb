```python
import numpy as np # 数値計算ライブラリ
import pandas as pd # データ解析ライブラリ
import matplotlib.pyplot as plt # グラフ描画ライブラリ
%matplotlib inline

# テストデータの生成
from sklearn.mixture import GaussianMixture

# k-means
from sklearn.cluster import KMeans

# クラスタリングの評価
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score

%pip install japanize_matplotlib
import japanize_matplotlib # 日本語フォント
```

##ガウス混合分布データの生成(K=3)

```python
# ガウス混合分布から疑似データを生成
np.random.seed(42)
K0=3
n_samples = 300

# 3つの構成要素のパラメータ設定
means = np.array([[-1, -1], [1, 3], [3, -1]])
covs = np.array([[[2.0, 0.5], [0.5, 2.0]],
                 [[1.5, -0.3], [-0.3, 2.0]],
                 [[0.8, 0.0], [0.0, 0.8]]])
weights = np.array([0.4, 0.3, 0.3])

# sklearnのGaussianMixtureインスタンスを作成し、パラメータを設定
gmm = GaussianMixture(n_components=3, covariance_type='full', random_state=42)
gmm.fit(np.random.randn(10, 2))  # パラメータを手動設定するために一度ダミーで初期化

# 事前に定義した平均、共分散、重みを設定
gmm.weights_ = weights
gmm.means_ = means
gmm.covariances_ = covs
gmm.precisions_cholesky_ = np.linalg.cholesky(np.linalg.inv(covs))

# ガウス混合モデルからサンプルを生成
X, y = gmm.sample(n_samples)
```

```python
# 生成したデータのプロット
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c='gray', alpha=0.6, edgecolors='k')
plt.title(f'生成されたガウス混合分布のデータ(K={K0})')
plt.xlabel("x1")
plt.ylabel("x2")
plt.grid(True)
plt.show()
```

##K-meansによるクラスタリングの可視化

### K=3

```python
# from sklearn.cluster import KMeans

# クラスターの個数の指定
K = 3

# KMeansモデルの定義と学習
kmeans = KMeans(n_clusters=K, n_init='auto', random_state=42)
kmeans.fit(X)
# 教師なし学習なので，別にpredictを行わない
# 訓練データXに対するlabelがつけられているのでそれを取り出す
# 予測クラスタラベルとクラスタ中心の取得
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# クラスタリング結果のプロット
plt.figure(figsize=(8, 6))

# クラスター数 K に応じて動的に色を生成（レインボーやtab10などのカラーマップを使用）
cormap = plt.colormaps.get_cmap('rainbow')
colors = [cormap(i / K) for i in range(K)]

for k in range(K):
    cluster_data = X[labels == k]
    plt.scatter(cluster_data[:, 0], cluster_data[:, 1], color=colors[k], alpha=0.5, label=f'クラスター {k}', edgecolors='k')

# クラスタ中心のプロット
plt.scatter(centroids[:, 0], centroids[:, 1], c='yellow', marker='X', s=200, edgecolors='black', linewidths=2, label='中心点')

plt.title(f'K-meansによるクラスタリング結果(K={K})')
plt.xlabel("x1")
plt.ylabel("x2")
plt.legend()
plt.grid(True)
plt.show()
```

### クラスタリング結果のエクスポート

```python
# カラムを x1, x2, labels として DataFrame を作成
df_results = pd.DataFrame({
    'x1': X[:, 0],
    'x2': X[:, 1],
    'cluster_label': labels
})

# 先頭5行を表示確認
display(df_results.head())

# 左側のファイル目次（/content/直下）にCSVを保存
output_path = '/content/kmeans_clustering_results.csv'
df_results.to_csv(output_path, index=False, encoding='utf-8-sig')
```

### クラスタリング結果の評価

```python
import pandas as pd
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score

# 評価用のKの値のリスト
k_values = range(2,7)
eval_results = []

for k_val in k_values:
    # 各KでKMeansを実行
    kmeans_model = KMeans(n_clusters=k_val, n_init='auto', random_state=42)
    cluster_labels = kmeans_model.fit_predict(X)

    # 各種評価指標の計算 (正解データ y を使わない評価)
#    sil = silhouette_score(X, cluster_labels)
    ch = calinski_harabasz_score(X, cluster_labels)
#    db = davies_bouldin_score(X, cluster_labels)

    eval_results.append({
        'K': k_val,
#        'シルエット係数 (↑高評価)': sil,
        'Calinski-Harabasz (CH) 指数 (↑高評価)': ch,
#        'Davies-Bouldin 指数 (↓高評価)': db
    })

# 結果をDataFrameにまとめて表示
df_eval = pd.DataFrame(eval_results)
display(df_eval)
```
