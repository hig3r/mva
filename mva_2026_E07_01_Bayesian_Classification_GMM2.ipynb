````python
import numpy as np # 数値計算ライブラリ
import pandas as pd # データ解析ライブラリ
import matplotlib.pyplot as plt # グラフ描画ライブラリ
%matplotlib inline

from sklearn.covariance import EmpiricalCovariance # scikit-learn の母共分散行列推定ライブラリ
from sklearn.metrics import confusion_matrix

import scipy.stats as stats # 統計ライブラリ

# %pip install japanize_matplotlib
# import japanize_matplotlib # 日本語フォント
````

次の母数は，課題 E07_02 では，CSVデータから推定する．[E03_03](https://learn.hig3.net/moodle/mod/assign/view.php?id=3062) 参照．

````python
# 最初から母分布が与えられているとする
pi=[0.3, 1-0.3] # yの事前分布
mu=[[-2,0],[1,1]] # 毋平均値
cov=[
    [[1,0.5],
     [0.5,2]],
    [[2,0.1],
     [0.1,3]]] #　母共分散行列（対称行列）大注意：標準偏差ではない
````

````python
rvx=[stats.multivariate_normal(mean=mu[0],cov=cov[0]),
     stats.multivariate_normal(mean=mu[1],cov=cov[1])]
````

[E06_03](https://learn.hig3.net/moodle/mod/url/view.php?id=3098)参照

````python
# 2次元データの散布図による可視化
plt.figure(figsize=(10, 8))
# 等高線を描画するためのグリッド作成
x1_min, x1_max =-10, 10
x2_min, x2_max = -10, 10
x1_grid, x2_grid = np.meshgrid(np.linspace(x1_min, x1_max, 200), np.linspace(x2_min, x2_max, 200))
pos = np.dstack((x1_grid, x2_grid))
# 各成分の確率密度（PDF）の計算
pdf0 = rvx[0].pdf(pos)
pdf1 = rvx[1].pdf(pos)
# 混合比率 pi=[0.6, 0.4] に基づく全体の確率密度
pdf_total = pi[0] * pdf0 + pi[1] * pdf1

# 等高線のプロット
# クラス0 (赤系), クラス1 (青系), 全体 (緑系)
contour0 = plt.contour(x1_grid, x2_grid, pdf0, levels=8, colors='red', alpha=0.6, linestyles='dashed')
contour1 = plt.contour(x1_grid, x2_grid, pdf1, levels=8, colors='blue', alpha=0.6, linestyles='dashed')
contour_total = plt.contour(x1_grid, x2_grid, pdf_total, levels=10, colors='green', alpha=0.8, linewidths=2)

plt.clabel(contour_total, inline=True, fontsize=8, fmt='%.3f')
````

````python
# この点を判別したい
x_test=[-1,0]
````

````python
# x = -1 における各クラスの確率密を計算
pdf_x=[ rvx[0].pdf(x_test),  rvx[1].pdf(x_test)]

# 事前確率 * 尤度 (事後分布postの分子 numerator の計算)
post_numerator=[ pi[0] * pdf_x[0], pi[1] * pdf_x[1]]

# 周辺分布密度（事後分布postの分母 denominator の計算）
post_denominator = post_numerator[0]+ post_numerator[1]

# 事後確率の計算
posty = post_numerator/post_denominator
print(f"--- x = {x_test} のときの事後確率 ---")
print(f"Class 0 の事後確率 P(y=0|x={x_test}): {posty[0]:.6f}")
print(f"Class 1 の事後確率 P(y=1|x={x_test}): {posty[1]:.6f}")

# クラスの決定
estimated_class = 0 if posty[0] > posty[1] else 1
print(f"\n推定されるクラス: Class {estimated_class}")
````
