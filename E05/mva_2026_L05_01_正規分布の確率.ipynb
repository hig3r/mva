## 標準正規分布 $Z\sim N(0,1^2)$

```python
rvz=stats.norm(loc=0,scale=1)
```

累積分布関数 $P(Z\leq z)=\Phi(z)=$.cdf(z)

$P(-1 \leq Z \leq 2)=\Phi(2)-\Phi(-1)$

```python
print(rvz.cdf(2)-rvz.cdf(-1))
```

$P(Z\geq 2)=P(2\leq Z \lt +\infty)=\Phi(+\infty)-\Phi(2)=1-\Phi(2)$

```python
print(1-rvz.cdf(2))
```

## 一般の正規分布$X\sim N(-3,2^2)$

標準化 $Z=\frac{X-(-3)}{2}\sim N(0,1^2)$

$P(-2 \leq X \leq 3)=P(\frac{-2-(-3)}{2}\leq Z \leq \frac{3-(-3)}{2})$

```python
print(rvz.cdf((3-(-3))/2)-rvz.cdf((-2-(-3))/2))
```
