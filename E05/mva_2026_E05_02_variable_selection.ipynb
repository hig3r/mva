```python
import numpy as np # 数値計算ライブラリ
import pandas as pd # データ解析ライブラリ
import matplotlib.pyplot as plt # グラフ描画ライブラリ
%matplotlib inline

from sklearn.linear_model import LinearRegression # scikit-learn の線形重回帰ライブラリ
from sklearn.model_selection import train_test_split # scikit-learn の訓練データとテストデータ分割ライブラリ
from sklearn.metrics import root_mean_squared_error, r2_score

# %pip install japanize_matplotlib
# import japanize_matplotlib # 日本語フォント
```

## 例

```python
#   「Pythonで理解する統計解析の基礎」 谷合廣紀，辻 真吾，技術評論社，2018. のデータを，dataframe形式で読み込み
df=pd.read_csv("https://github.com/ghmagazine/python_stat_sample/raw/master/data/ch12_scores_reg.csv")
df.head()
```

ここでは，期末テストを目的変数，小テストと睡眠時間を説明変数候補にします．

```python
y=df['期末テスト']
```

この分析では，訓練データとテストデータへの分割は行わず，全データを使用し，自由度調整済み決定係数で変数選択します．

```python
# 線形重回帰モデルを使用します
model=LinearRegression()
```

### 目的変数={小テスト}，単回帰

```python
X1=df.drop(['期末テスト','通学方法','睡眠時間'], axis=1)
n=X1.shape[0]
p=X1.shape[1]
model.fit(X1,y)
print(X1.columns.tolist())
print(model.intercept_)
print(model.coef_)
print(n)
print(p)
print(model.intercept_) # 切片
print(model.coef_) # 回帰係数
print(model.score(X1,y)) # 決定係数
print( 1-(1-model.score(X1,y))*(n-1)/(n-p-1)) # 自由度調整済み決定係数 R^*2
```

### 目的変数={睡眠時間}，単回帰

```python
X2=df.drop(['期末テスト','通学方法','小テスト'], axis=1)
n=X2.shape[0]
p=X2.shape[1]
model.fit(X2,y)
print(X1.columns.tolist())
print(model.intercept_)
print(model.coef_)
print(n)
print(p)
print(model.intercept_) # 切片
print(model.coef_) # 回帰係数
print(model.score(X2,y)) # 決定係数
print( 1-(1-model.score(X2,y))*(n-1)/(n-p-1)) # 自由度調整済み決定係数 R^*2
```

### 目的変数={小テスト,睡眠時間}，重回帰

```python
X12=df.drop(['期末テスト','通学方法'], axis=1)
n=X12.shape[0]
p=X12.shape[1]
model.fit(X12,y)
print(X12.columns.tolist())
print(model.intercept_)
print(model.coef_)
print(n)
print(p)
print(model.intercept_) # 切片
print(model.coef_) # 回帰係数
print(model.score(X12,y)) # 決定係数
print( 1-(1-model.score(X12,y))*(n-1)/(n-p-1)) # 自由度調整済み決定係数 R^*2
```

結果から，3個の変数選択（説明変数として採用する変数の組み合わせ)のうち，どれが最適か，$R^{\ast 2}$ に基づいて判定できたはず．
