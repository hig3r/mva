````python
import numpy as np # 数値計算ライブラリ
import pandas as pd # データ解析ライブラリ
import matplotlib.pyplot as plt # グラフ描画ライブラリ
%matplotlib inline

from sklearn.linear_model import LinearRegression # scikit-learn の線形重回帰ライブラリ
from sklearn.model_selection import train_test_split # scikit-learn の訓練データとテストデータ分割ライブラリ
from sklearn.metrics import root_mean_squared_error, r2_score

# %pip install japanize_matplotlib
# import japanize_matplotlib # 日本語フォント
````

## 例

````python
#   「Pythonで理解する統計解析の基礎」 谷合廣紀，辻 真吾，技術評論社，2018. のデータを，dataframe形式で読み込み
df=pd.read_csv("https://github.com/ghmagazine/python_stat_sample/raw/master/data/ch12_scores_reg.csv")
df.head()
````

徒歩(1)かそれ以外(0)かダミー変数として追加します．2値のカテゴリ変数はこのようなダミー変数にしてよいですが，
(順序のない)3値以上のカテゴリ変数を，バス=0,自転車=1,徒歩=2などとしてダミー変数かするのは正しくありません．

````python
df['徒歩'] = (df['通学方法'] == '徒歩').astype(int)
display(df.head())
````

上のコードの説明です．`df['徒歩']=ナントカ`，で，dfに新しい列(Series)を追加して代入します
`(df['通学方法']=='徒歩')`は，行ごとに真偽を調べた，値が true or false の，長さnのSeriesです．それをas typeでintにタイプキャストします．

Cで言ったら下に相当します．
```C
short trueorfalse=(x==1);
int dummy=(int)trueorfalse;
```

下では，既存の説明変数の積やべき乗で新しい説明変数を追加します．

````python
df['x6'] = df['小テスト'] * df['睡眠時間'] # Series(配列)の成分ごとの積
df['x7'] = df['睡眠時間'] ** 2 # Series(配列)の成分ごとの二乗
display(df.head())
````

目的変数は期末テストのままとして，
説明変数は使うものだけを残し，説明変数X, 目的変数yを作ります．

````python
X=df.drop(["通学方法","睡眠時間","期末テスト"],axis=1)
y=df["期末テスト"] # 目的変数
print(X)
print(y)
````

````python
# 線形重回帰モデルを使用します
model=LinearRegression()
````

````python
# X,yをそれぞれ訓練データとテストデータに，0.8:0.2の比率で分割する
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
# 先頭でsklearn.model_selectionからライブラリをimportしている
# 右辺はリストを返す．
# 左辺はリスト[X_train, y_train,...]に代入するのと同じこと

# train 訓練
# validation 検証（今回は使っていない）
# test テスト
# predict, pred 予測
# fit 推定
````

````python
print(X.shape)
print(X_train.shape)
print(X_test.shape)

print(y.shape)
print(y_train.shape)
print(y_test.shape)
````

````python
model.fit(X_train,y_train) # 母数推定 = 訓練は訓練データだけで行う．
````

````python
print(X_train.columns.tolist())
print(model.intercept_)
print(model.coef_)

print(model.score(X_train,y_train)) # 決定係数 R^2

n=X_train.shape[0]
p=X_train.shape[1]
print( 1-(1-model.score(X_train,y_train))*(n-1)/(n-p-1)) # 自由度調整済み決定係数 R^*2
````

````python
# 同じ量を別の方法で計算します決定係数はこのように計算することもできる
y_pred=model.predict(X_train) #予測値
print(r2_score(y_train,y_pred)) #実測値の，予測からの誤差でR^2を計算
````

機械学習ののりでは，性能評価のための決定係数は，訓練データとは別のテストデータから計算する．少し悪化することが多い．

````python
# model は訓練済み
print(model.score(X_test,y_test)) # 決定係数 R^2
# テストデータについては自由度調整済み決定係数の算出は不要．自由度で調整する，テストデータを別にする，は，同じ目的の作業なので，二重にはやらない
````

````python
# 決定係数はこのように計算することもできる
y_pred=model.predict(X_test) #予測値
print(r2_score(y_test,y_pred)) #実測値の，予測からの誤差でR^2を計算
````
