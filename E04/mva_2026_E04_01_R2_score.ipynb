```python
import numpy as np # 数値計算ライブラリ
import pandas as pd # データ解析ライブラリ
import matplotlib.pyplot as plt # グラフ描画ライブラリ
%matplotlib inline

from sklearn.linear_model import LinearRegression # scikit-learn の線形重回帰ライブラリ
from sklearn.model_selection import train_test_split # scikit-learn の訓練データとテストデータ分割ライブラリ
from sklearn.metrics import root_mean_squared_error, r2_score

!pip install japanize_matplotlib
import japanize_matplotlib # 日本語フォント

from mpl_toolkits.mplot3d import Axes3D # 3Dプロット用のインポート
import plotly.graph_objects as go # 3次元インタラクティブプロット
```

```python
#   「Pythonで理解する統計解析の基礎」 谷合廣紀，辻 真吾，技術評論社，2018. のデータを，dataframe形式で読み込み
df=pd.read_csv("https://github.com/ghmagazine/python_stat_sample/raw/master/data/ch12_scores_reg.csv")
df
```

```python
y=df["期末テスト"] # 目的変数
X=df.drop(["通学方法","期末テスト"],axis=1)
X
```

```python
model=LinearRegression()
```

X,yを訓練データとテストデータに，0.8:0.2の比率で分割する
右辺はリストを返す．
左辺はリスト[X_train, y_train,...]を書いてもよい

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
# 先頭でライブラリをimportしている
```

```python
print(X.shape)
print(y.shape)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
model.fit(X_train,y_train) # 母数推定 = 訓練は訓練データだけで行う．
```

```python
print(X.columns.tolist())
print(model.coef_)
print(model.intercept_)

print(model.score(X_train,y_train)) # 決定係数 R^2

n=X_train.shape[0]
p=X_train.shape[1]
print( 1-(1-model.score(X_train,y_train))*(n-1)/(n-p-1)) # 自由度調整済み決定係数 R^*2
```

```python
# 決定係数はこのように計算することもできる
y_pred=model.predict(X_train)
r2=r2_score(y_train,y_pred)
r2
```

機械学習ののりでは，決定係数，自由度調整済み決定係数は，テストデータから計算するのが正しい

```python
print(y_pred)
print(y_test)
```

```python
# model は訓練済み
print(model.score(X_test,y_test)) # 決定係数 R^2
y_pred=model.predict(X_test)
print(r2_score(y_test,y_pred))
```
