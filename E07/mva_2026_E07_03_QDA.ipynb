## 可視化して様子を見る

```python
plt.figure(figsize=(8, 5))
plt.hist(df['X'], bins=30, density=True, alpha=0.6, color='g', edgecolor='black')
plt.title('xのヒストグラム')
plt.xlabel('x')
plt.ylabel('密度')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

2つのガウス分布の混合と見てもおかしくないかも？ QDAを使おう！

QDA=Quadratic_Discriminant_Analysis は，母数の推定と尤度の比較とをまとめて行う

## QDA

```python
x_test = -1
```

```python
# 1. 特徴量とターゲットの準備
X_train = df[['X']]
y_train = df['y']

# 2. QDAモデルのインスタンス生成と適合 (データの事前確率を自動推定)
qda = QuadraticDiscriminantAnalysis() # モデル
qda.fit(X_train, y_train) # 訓練=母数の推定

```

```python
# 3. x_test = -1 の所属確率 (事後確率) を推定
# 訓練データと同じ特徴量名 'X' を持つ DataFrame として作成します
X_test = pd.DataFrame([[x_test]], columns=['X'])
probs_qda = qda.predict_proba(X_test)[0] # 確率
pred_class_qda = qda.predict(X_test)[0] # クラス

print("--- QDA (二次判別分析) による予測結果 ---")
print(f"Class 0 の事後確率 P(y=0|x={x_test}): {probs_qda[0]:.6f}")
print(f"Class 1 の事後確率 P(y=1|x={x_test}): {probs_qda[1]:.6f}")
print(f"推定されるクラス: Class {pred_class_qda}")
```
