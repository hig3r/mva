## 説明

特定の分布族ではないけれど，モデルのようなもの

```python
model = EmpiricalCovariance()
```

```python
# UCIのデータを使わせていただく．
# ヘッダー（列名）がついていないので，この段階で5個つける
df = pd.read_csv(
    "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data",
    header=None, # 元データに見出し行がない前提で
    names=[
        "sepal_length", "sepal_width",
        "petal_length", "petal_width",
        "species",
    ], # この見出し行をつける
)
print(df.shape)
```

```python
# 一部分だけ取り出す（課題ではこの処理はしないので，理解せず無視して良い）
df2 = df.loc[
    df["species"].eq("Iris-versicolor"),
    ["petal_length", "petal_width"],
]

X = df2.to_numpy() # NumPy配列に変換

print(df2.shape)  # (50, 2)
print(X.shape)    # (50, 2)
```

母数の推定を行う．結果は，オブジェクトのプロパティとして保存される

```python
model.fit(X)
```

母数の推定値の取り出し

```python
print(model.location_)       # 母平均値ベクトルの推定値
print(model.covariance_)     # 母共分散行列の推定値
```

これらの推定値と同じ母数を持つ多変量正規分布

```python
rvy=stats.multivariate_normal(mean=np.array(model.location_),cov=np.array(model.covariance_))
```
