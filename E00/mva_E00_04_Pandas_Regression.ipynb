## ガリガリ君 df1

Pythonで多変量データを扱うライブラリ，Pandasでは，ローカルやネット上の`.csv`ファイルを読み込むことができます．
それを，列の名前を持った`DataFrame`として持つことができます．

```python
# Webから高橋さんのゴリゴリくんデータを読み込み，データフレーム df1 とする
df1 = pd.read_csv('https://www-tlab.math.ryukoku.ac.jp/~takataka/course/Data/ex08gorigori.csv')
```

```python
df1
```

```python
plt.scatter(x=df1['気温'], y=df1['アイス売上数']) # scatter plot=散布図
plt.xlabel('気温')
plt.ylabel('アイス売上数')
plt.title('気温とアイス売上数の散布図')
plt.show()
```

```python
# 標本平均値，不偏標本分散，不偏標本共分散を求める
x_mean=df1['気温'].mean()
y_mean=df1['アイス売上数'].mean()
sxx=df1['気温'].var()
syy=df1['アイス売上数'].var()
# sxy=df1['気温'].cov(df1['アイス売上数'])
sxy=df1.cov()['気温']['アイス売上数'] 
```

```python
beta1=sxy/sxx
beta1
```

```python
beta0=y_mean-beta1*x_mean
beta0
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(x=df1['気温'], y=df1['アイス売上数'], label='実測値')

# 回帰直線の計算
y_pred = beta1 * df1['気温'] + beta0
plt.plot(df1['気温'], y_pred, color='red', label=f'回帰直線 (y = {beta1:.2f}x + {beta0:.2f})')

plt.xlabel('気温')
plt.ylabel('アイス売上数')
plt.title('気温とアイス売上数の散布図と回帰直線')
plt.legend()
plt.grid(True)
plt.show()
```
