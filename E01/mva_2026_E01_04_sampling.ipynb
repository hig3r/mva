## 1次元の連続型確率変数
連続型確率変数$X$を考える. いま例として, **正規分布**に従う場合, 具体的には$X\sim\mathrm{N}(\mu,\sigma^2), \mu=5, \sigma=3$ の場合を考える.

$$
f(x)=\frac{1}{(2\pi \sigma^2)^{1/2}} e^{-\frac{(x-\mu)^2}{\sigma^2}}
$$

#### scipy
確率変数$Z\sim\mathrm{N}(m,s^2)$ は, `stats.norm(loc=m,scale=s)`で作れる. `loc`=location=位置=(この分布の場合は)母平均値, `scale`=大きさ=ちらばり=(この場合は)母標準偏差.

この確率変数に対して，`.rvs(size=n)`を実行することで，サイズ`n`の標本がリストとして得られる．


```python
mu=5 #母平均値
sigma=3 #母標準偏差
rvz=stats.norm(loc=mu,scale=sigma)
s=rvz.rvs(size=n)
print(s)
```

```python
実は，いったん`rvz`を定義しなくても次でもよかった
```

```python
stats.norm(loc=mu,scale=sigma).rvs(size=n)
```

`s`に`datax`という名前（自由につけられる）をつけて，表`dfn`に格納する．

```python
dfn=pd.DataFrame(data={'datax':s})
dfn
```

`{'キー':値, 'キー':値,...}` は，実はPythonのデータ構造のひとつで，dictionaryと呼ばれる．

全部を1行でやるなら，

```python
dfn=pd.DataFrame(data={'datax':stats.norm(loc=mu,scale=sigma).rvs(size=n)})
```

```python
# 基本統計量
# describe=記述する
# mean=標本平均値, std=標本標準偏差=標本分散の平方根, count=標本サイズ
dfn.describe()
```

## 2次元の連続型確率分布
連続型確率変数$X,Y$が**独立**な場合を考える. いま例として, $X\sim\mathrm{N}(2,3^2), Y\sim\mathrm{N}(3,5^2)$の場合を考える.

#### pandas
横の並びは 行 row
縦の並びは 列 column
線形代数の行列の同じ呼び名ですね


```python
ux=stats.norm(loc=2,scale=3).rvs(size=n)
uy=stats.norm(loc=3,scale=5).rvs(size=n)
dfn2=pd.DataFrame(data={'datax':ux,'datay':uy}) # (表)PandasのDataFrameに変数名をつけて代入
dfn2 # 表示
```

```python
p=dfn2.plot.scatter(x='datax',y='datay',s=0.1) # pandas の散布図(scatter plot), x= は横軸の変数指定. 'datax' は表の見出し. s=ドットのサイズ
p.set_xlabel('x') # やっぱり軸ラベルは 'x' で
p.set_aspect('equal') # 縦横比 aspect ratio
```

```python
#周辺分布
p=dfn2['datay'].plot.hist()
p.set_xlabel("datay")
p.set_title("yの周辺分布の標本のヒストグラム")
p
```

## ふたたび1次元の連続型確率変数

```python
# 連続型一様分布 U(c,d)
c=3
d=5
rvu=stats.uniform(loc=c,scale=d-c)
rvu.rvs(size=n)
```
