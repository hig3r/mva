## 解析の一般的な流れ
1. 標本を配列として取得する
    1. 確率統計ののりで特定の分布から抽出するときは, `stats.分布名.rvs` で（今週）

### 一般的な流れの準備
シードが同じなら, ランダムでも同じ標本が抽出される. デバッグ上有効.   
サンプルサイズを固定. nに代入

```python
np.random.seed(1637)
n=1000
```

## 1次元の連続型確率変数
連続型確率変数$X$を考える. いま例として, **正規分布**に従う場合, 具体的には$X\sim\mathrm{N}(\mu,\sigma^2), \mu=5, \sigma=3$ の場合を考える.

$$
f(x)=\frac{1}{(2\pi \sigma^2)^{1/2}} e^{-\frac{(x-\mu)^2}{\sigma^2}}
$$

#### scipy
$\mathrm{N}(m,s^2)$ は, `stats.norm.rvs(loc=m,scale=s,size=n)`で生成できる. loc=location=位置=(この分布の場合は)母平均値, scale=大きさ=ちらばり=(この場合は)母標準偏差.

#### Python
`{'キー':値, 'キー':値,...}` はPythonのdictionary.

```python
mu=5 #母平均値
sigma=3 #母標準偏差
dx=stats.norm.rvs(loc=mu,scale=sigma,size=n)
```

```python
dfn=pd.DataFrame(data={'datax':dx})
dfn
```

```python
# 基本統計量
# describe=記述する
# mean=標本平均値, std=標本標準偏差=標本分散の平方根, count=標本サイズ
dfn.describe()
```

```python
p=dfn.plot.hist(bins=30)
p.set_xlabel("x")
p.set_title("N(5,3^2)の標本のヒストグラム")
p
```

## 2次元の連続型確率分布
連続型確率変数$X,Y$が**独立**な場合を考える. いま例として, $X\sim\mathrm{N}(2,3^2), Y\sim\mathrm{N}(3,5^2)$の場合を考える.

#### pandas
横の並びは 行 row
縦の並びは 列 column
線形代数の行列の同じ呼び名ですね


```python
ux=stats.norm.rvs(loc=2,scale=3,size=n)
uy=stats.norm.rvs(loc=3,scale=5,size=n)
dfcu=pd.DataFrame(data={'datax':ux,'datay':uy}) # (表)PandasのDataFrameに変数名をつけて代入
dfcu # 表示
```

```python
#基本統計量
dfcu.describe()
```

```python
s=dfcu.plot.scatter(x='datax',y='datay',s=0.1) # pandas の散布図(scatter plot), x= は横軸の変数指定. 'datax' は表の見出し. s=ドットのサイズ
s.set_xlim(-12,12) # 軸の範囲(limit)
s.set_ylim(-12,12)
s.set_xlabel('x') # やっぱり軸ラベルは 'x' で
s.set_aspect('equal') # 縦横比 aspect ratio
```
